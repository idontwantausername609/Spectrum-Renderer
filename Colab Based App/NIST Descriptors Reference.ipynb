# NIST Descriptor Reference

This notebook defines all NIST intensity descriptors and specifies how to handle each one in spectral analysis.

## NIST Descriptor Definitions & Actions

| Descriptor | Meaning | Visual Action | Intensity Multiplier | Width Multiplier | Include? |
|---|---|---|---|---|---|
| (none) | Normal line | Render as a single line | 1.0 | 1.0 | ✓ |
| `*` | Intensity shared by several lines (typically multiply classified) | Keep as a normal line, but note shared intensity | 1.0 | 1.0 | ✓ |
| `:` | Observed value given is actually the rounded Ritz value | Keep as a normal line with lower confidence in wavelength | 1.0 | 1.0 | ✓ |
| `-` | Somewhat lower intensity than the value given | Dim the line slightly | 0.8 | 1.0 | ✓ |
| `a` | Observed in absorption | Dim strongly; consider marking as absorption | 0.5 | 1.0 | ✓ |
| `b` | Band head | Render as an edge/band feature, not a thin isolated line | 1.0 | 1.2 | ✓ |
| `bl` | Blended with another line; wavelength and intensity may be affected | Merge into a blended feature | 1.5 | 1.25 | ✓ (⚠️ uncertain) |
| `B` | Line or feature having large width due to autoionization broadening | Render as a broad line | 1.5 | 1.8 | ✓ |
| `c` | Complex line | Slightly broaden; keep as a single feature unless data resolve components | 1.0 | 1.1 | ✓ |
| `d` | Diffuse line | Slightly broaden | 1.0 | 1.1 | ✓ |
| `D` | Double line | Render as two nearby components, not a single broader line | 1.0 | 1.0 | ✓ |
| `E` | Broad due to overexposure in the quoted reference | Render as broad and slightly dimmer | 0.9 | 1.3 | ✓ (⚠️) |
| `f` | Forbidden line | Render as a narrow line; keep full intensity unless other data say otherwise | 1.0 | 1.0 | ✓ |
| `g` | Transition involving a level of the ground term | Render as a normal line | 1.0 | 1.0 | ✓ |
| `G` | Line position roughly estimated | Render as a normal line, but mark wavelength as uncertain | 1.0 | 1.0 | ? |
| `H` | Very hazy line | Render as a broad, low-confidence feature | 0.7 | 1.5 | ✓ (⚠️) |
| `h` | Hazy line (same as "diffuse") | Slightly broaden | 1.0 | 1.1 | ✓ |
| `hfs` | Line has hyperfine structure | Render as an unresolved multiplet cluster | 1.0 | 1.35 | ✓ |
| `i` | Identification uncertain | Dim the line; keep normal width | 0.6 | 1.0 | ✓ (⚠️) |
| `j` | Wavelength smoothed along isoelectronic sequence | Render as a normal line | 1.0 | 1.0 | ✓ |
| `l` | Shaded to longer wavelengths; may look like a "1" at the end of the number | Render as a normal line | 1.0 | 1.0 | ✓ |
| `m` | Masked by another line (no wavelength measurement) | **SKIP** — no valid wavelength | 0.0 | 0.0 | ✗ |
| `p` | Perturbed by a close line; wavelength and intensity may be affected | Slightly broaden and dim | 0.8 | 1.2 | ✓ (⚠️) |
| `q` | Asymmetric line | Render as a normal line | 1.0 | 1.0 | ✓ |
| `r` | Easily reversed line | Render as a normal line | 1.0 | 1.0 | ✓ |
| `s` | Shaded to shorter wavelengths | Render as a normal line | 1.0 | 1.0 | ✓ |
| `t` | Tentatively classified line | Dim the line; keep normal width | 0.7 | 1.0 | ✓ (⚠️) |
| `u` | Unresolved from a close line | Dim and slightly broaden; may represent a hidden cluster | 0.75 | 1.2 | ✓ (⚠️) |
| `w` | Wide line | Render as a broad line | 1.0 | 1.8 | ✓ |
| `x` | Extrapolated wavelength | Render as a normal line, but flag wavelength as extrapolated | 1.0 | 1.0 | ✓ (⚠️) |

### Legend
- **✓** = Include in visualization
- **✓ (⚠️)** = Include but flag as uncertain/modified
- **?** = Include with caution
- **✗** = Exclude
- **Intensity Multiplier** = Multiply the reported intensity by this value
- **Width Multiplier** = Multiply the rendered line width by this value

## Current Implementation Strategy

### Descriptors with Intensity Corrections:
- **`-`** (somewhat lower): 0.8x (reduce by 20%)
- **`a`** (absorption): 0.5x (reduce by 50%)
- **`p`** (perturbed): 0.8x (reduce by 20%)
- **`H`** (very hazy): 0.7x (reduce by 30%)
- **`t`** (tentative): 0.7x (reduce by 30%)
- **`u`** (unresolved): 0.75x (reduce by 25%)
- **`E`** (overexposed): 0.9x (reduce by 10%)
- **`i`** (uncertain): 0.6x (reduce by 40%)

### Descriptors with Width or Shape Changes:
- **`B`** (autoionization broadening): broad line
- **`w`** (wide): broad line
- **`H`** (very hazy): broad, low-confidence line
- **`h`** (hazy/diffuse): slightly broadened line
- **`b`** (band head): edge-like or band feature
- **`bl`** (blended): blended feature
- **`c`** and **`d`**: slightly broadened complex/diffuse lines
- **`hfs`** and **`u`**: unresolved clusters or multiplet-like features
- **`D`** (double line): two nearby components, not a single fat line

### Descriptors with No Correction (Use 1.0x):
All others: `*`, `:`, `f`, `g`, `G`, `j`, `l`, `q`, `r`, `s`, `x`

### Descriptors to Skip:
- **`m`** (masked): Skip entirely — no valid wavelength measurement

## Descriptor Grouping by Reliability

### High Confidence (no correction needed)
Normal lines, forbidden lines, ground-term lines, smoothed lines, shaped lines, and easily-reversed lines.

### Medium Confidence (apply intensity correction or mild broadening)
Somewhat lower (`-`), perturbed (`p`), overexposed (`E`), unresolved (`u`), and band-like or blended features (`b`, `bl`, `c`, `d`, `h`, `hfs`).

### Low Confidence (wavelength uncertain or identification uncertain)
Roughly estimated position (`G`), extrapolated (`x`), shared intensity (`*`), tentative (`t`), and uncertain identification (`i`).

### Excluded (unusable)
Masked lines (`m`) — no wavelength measurement available
